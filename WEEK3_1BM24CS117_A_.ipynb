```python
import heapq

class Node:
    def __init__(self, state, parent=None, action=None, g=0, h=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.g = g
        self.h = h
        self.f = g + h

    def __lt__(self, other):
        return self.f < other.f

class Problem:
    def __init__(self, initial_state, goal_state):
        self.initialState = initial_state
        self.goalState = goal_state

    def goalTest(self, state):
        return state == self.goalState

    def actions(self, state):
        blank_idx = state.index(0)
        row, col = divmod(blank_idx, 3)

        valid_actions = []
        if row > 0: valid_actions.append('Up')
        if row < 2: valid_actions.append('Down')
        if col > 0: valid_actions.append('Left')
        if col < 2: valid_actions.append('Right')
        return valid_actions

    def result(self, state, action):
        blank_idx = state.index(0)
        next_state = list(state)

        if action == 'Up': delta = -3
        elif action == 'Down': delta = 3
        elif action == 'Left': delta = -1
        elif action == 'Right': delta = 1

        next_idx = blank_idx + delta
        next_state[blank_idx], next_state[next_idx] = next_state[next_idx], next_state[blank_idx]
        return tuple(next_state)

def misplaced_tiles(state, goal_state):
    count = 0
    for idx, value in enumerate(state):
        if value != 0 and value != goal_state[idx]:
            count += 1
    return count

def childNode(problem, parent, action):
    next_state = problem.result(parent.state, action)
    g_prime = parent.g + 1
    h_prime = misplaced_tiles(next_state, problem.goalState)
    return Node(state=next_state, parent=parent, action=action, g=g_prime, h=h_prime)

def solution_path(node):
    path = []
    while node.parent is not None:
        path.append((node.action, node.state, node.g, node.h))
        node = node.parent
    return path[::-1]

def a_star_search(problem):
    start_node = Node(state=problem.initialState, g=0, h=misplaced_tiles(problem.initialState, problem.goalState))
    frontier = []
    heapq.heappush(frontier, start_node)
    explored = set()

    while True:
        if not frontier:
            return "failure"

        n = heapq.heappop(frontier)

        if problem.goalTest(n.state):
            return solution_path(n)

        if n.state not in explored:
            explored.add(n.state)

            for a in problem.actions(n.state):
                n_prime = childNode(problem, n, a)

                if n_prime.state not in explored:
                    heapq.heappush(frontier, n_prime)

def get_state_input(prompt):
    print(prompt)
    state = []
    for i in range(3):
        row = list(map(int, input(f"Enter row {i+1} separated by spaces (use 0 for blank): ").split()))
        state.extend(row)
    return tuple(state)

initial_board = get_state_input("--- Enter Initial State ---")
goal_board = get_state_input("--- Enter Goal State ---")

puzzle_problem = Problem(initial_board, goal_board)
result = a_star_search(puzzle_problem)

if result != "failure":
    print(f"\nSolved successfully in {len(result)} moves!\n")
    print("Initial State:")
    for row in range(0, 9, 3):
        print(list(initial_board[row:row+3]))
    print(f"g = 0, h = {misplaced_tiles(initial_board, goal_board)}\n")

    for step, board, g, h in result:
        print(f"Move: {step}")
        for row in range(0, 9, 3):
            print(list(board[row:row+3]))
        print(f"g = {g}, h = {h}, f(g+h) = {g+h}\n")
else:
    print("\nFailed to find a solution.")

```
