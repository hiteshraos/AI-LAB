```python
# Goal state configuration
GOAL_STATE = (1, 2, 3,
              8, 0, 4,
              7, 6, 5)

# Moves map based on the 0-indexed position of the blank tile (0)
# Format: index -> list of indices it can swap with (Up, Down, Left, Right)
MOVES = {
    0: [1, 3],       1: [0, 2, 4],    2: [1, 5],
    3: [0, 4, 6],    4: [1, 3, 5, 7], 5: [2, 4, 8],
    6: [3, 7],       7: [4, 6, 8],    8: [5, 7]
}

def get_neighbors(state):
    """Generates next valid states by moving the blank tile (0)."""
    neighbors = []
    blank_idx = state.index(0)

    for next_idx in MOVES[blank_idx]:
        # Swap blank tile with adjacent tile
        new_state = list(state)
        new_state[blank_idx], new_state[next_idx] = new_state[next_idx], new_state[blank_idx]
        neighbors.append(tuple(new_state))

    return neighbors

def depth_limited_search(state, depth, path, visited_path):
    """Performs a depth-limited DFS tracking the current recursion path."""
    if state == GOAL_STATE:
        return path

    if depth <= 0:
        return None

    # Add to current path to prevent cycles in this branch
    visited_path.add(state)

    for neighbor in get_neighbors(state):
        if neighbor not in visited_path:
            result = depth_limited_search(neighbor, depth - 1, path + [neighbor], visited_path)
            if result is not None:
                return result

    # Backtrack
    visited_path.remove(state)
    return None

def iddfs(initial_state, max_depth=10):
    """Executes Iterative Deepening DFS up to a maximum depth limit."""
    print("Searching for solution...")
    for depth in range(max_depth + 1):
        # Tracking states along the current recursion branch to avoid loops
        visited_path = set()
        result = depth_limited_search(initial_state, depth, [initial_state], visited_path)

        if result is not None:
            return result, depth

    return None, max_depth

def print_board(state):
    """Helper to format and print the 3x3 puzzle board."""
    for i in range(0, 9, 3):
        print(f" {state[i]} {state[i+1]} {state[i+2]} ")
    print("-" * 11)

# --- Execution ---
if __name__ == "__main__":
    # A solvable puzzle configuration (Requires 4 moves)
    initial_puzzle = (2, 8, 3,
                      1, 6, 4,
                      7, 0, 5)

    solution_path, depth_found = iddfs(initial_puzzle)

    if solution_path:
        print(f"\nSuccess! Solution found at Depth: {depth_found}")
        print("Steps:")
        for step, board in enumerate(solution_path):
            print(f"Step {step}:")
            print_board(board)
    else:
        print("No solution found within depth limits.")
```
