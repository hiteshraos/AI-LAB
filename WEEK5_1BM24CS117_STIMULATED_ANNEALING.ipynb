```python
import math
import random

class SimulatedAnnealing8Queens:
    def __init__(self, initial_state):
        # 1. State Representation: 1D array of length 8 where board[col] = row
        self.initial_state = list(initial_state)

    def calculate_cost(self, state):
        """2. Cost Function (Objective): Count all unique pair conflicts."""
        conflicts = 0
        n = len(state)
        for c1 in range(n):
            for c2 in range(c1 + 1, n):
                r1 = state[c1]
                r2 = state[c2]

                # Check for Row Collision or Diagonal Collision
                if r1 == r2 or abs(r1 - r2) == abs(c1 - c2):
                    conflicts += 1
        return conflicts

    def get_random_neighbor(self, state):
        """3. Neighbor Generation: Pick a random column and change its row value."""
        neighbor = list(state)
        col = random.randint(0, 7)

        # Pick a different random row from 0 to 7
        current_row = state[col]
        possible_rows = [r for r in range(8) if r != current_row]
        new_row = random.choice(possible_rows)

        neighbor[col] = new_row
        return col, new_row, neighbor

    def run_simulation(self, initial_temp=10.0, alpha=0.995, max_iterations=10000):
        current_board = list(self.initial_state)
        current_cost = self.calculate_cost(current_board)
        T = initial_temp
        iteration = 0

        print("\n=== STARTING SIMULATED ANNEALING ===")
        print(f"Initial Configuration: {current_board} | Cost: {current_cost} conflicts\n")

        while T > 0.0001 and current_cost > 0 and iteration < max_iterations:
            iteration += 1
            col, new_row, proposed_board = self.get_random_neighbor(current_board)
            proposed_cost = self.calculate_cost(proposed_board)

            # 4. Acceptance Criterion: ΔE = Cost_new - Cost_current
            delta_E = proposed_cost - current_cost

            # Case 1: ΔE < 0 (Better Move) -> Always Accept
            if delta_E < 0:
                current_board = proposed_board
                current_cost = proposed_cost
                print(f"[Iter {iteration:03d}] Better Move -> Board: {current_board} | Cost: {current_cost} | T: {T:.4f}")

            # Case 2: ΔE >= 0 (Worse Move) -> Accept Probabilistically
            else:
                p = math.exp(-delta_E / T)
                if random.random() < p:
                    current_board = proposed_board
                    current_cost = proposed_cost
                    print(f"[Iter {iteration:03d}] Worse Move Accepted (P={p:.3f}) -> Board: {current_board} | Cost: {current_cost} | T: {T:.4f}")

            # 5. Cooling Schedule: T_(k+1) = T_k * alpha
            T *= alpha

        print("\n=== CONVERGENCE SUMMARY ===")
        if current_cost == 0:
            print(f"🎉 SUCCESS! Found goal state: {current_board}")
        else:
            print(f"🛑 Stopped at Local Minimum: {current_board} | Remaining Cost: {current_cost}")

        self.print_board(current_board)

    def print_board(self, state):
        """Draws a visual text-based 8x8 chessboard matrix."""
        print("\nVisual Board Layout:")
        for row in range(8):
            line = ""
            for col in range(8):
                line += " Q " if state[col] == row else " . "
            print(line)

# --- Interactive Terminal Execution ---
if __name__ == "__main__":
    print("Name: HITESH")
    print("USN: 1BM24CS117")
    print("====================================================")
    print("        8-Queens Simulated Annealing Solver         ")
    print("====================================================")
    print("Provide your initial board as 8 row positions (0 to 7)")
    print("separated by spaces. Example layout: 0 1 2 3 4 5 6 7")
    print("----------------------------------------------------")

    # Prompt the user for live input
    user_input = input("Enter initial state: ").strip()

    try:
        # Convert text input into an array of integers
        initial_state = [int(x) for x in user_input.split()]

        # Rigorous input validation checks
        if len(initial_state) != 8:
            print("\n❌ Error: You must enter exactly 8 numbers.")
        elif any(x < 0 or x > 7 for x in initial_state):
            print("\n❌ Error: All row numbers must be between 0 and 7.")
        else:
            # Validated -> Run simulation
            solver = SimulatedAnnealing8Queens(initial_state)
            solver.run_simulation()

    except ValueError:
        print("\n❌ Error: Invalid formatting. Please enter integers separated by spaces only.")
```
