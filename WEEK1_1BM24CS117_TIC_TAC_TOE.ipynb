```python
import random

# Initialize the 3x3 game board
def print_board(board):
    print("\n")
    for i in range(3):
        print(" | ".join(board[i]))
        if i < 2:
            print("---------")
    print("\n")

# Check if a player has won
def check_win(board, player):
    # Check rows, columns, and diagonals
    for i in range(3):
        if all(board[i][j] == player for j in range(3)) or \
           all(board[j][i] == player for j in range(3)):
            return True
    if all(board[i][i] == player for i in range(3)) or \
       all(board[i][2 - i] == player for i in range(3)):
        return True
    return False

# Check if the board is completely full (tie)
def check_tie(board):
    return all(board[i][j] != " " for i in range(3) for j in range(3))

# Main game function
def play_game():
    board = [[" " for _ in range(3)] for _ in range(3)]
    current_player = "X"  # X is Human, O is Machine
    print("HITESH RAO S DALAVE , 1BM24CS117")
    print("Welcome to Tic-Tac-Toe! You are 'X' and the computer is 'O'.")

    while True:
        print_board(board)

        if current_player == "X":
            # Human turn
            try:
                row = int(input("Enter row (0, 1, or 2): "))
                col = int(input("Enter column (0, 1, or 2): "))

                if row not in [0, 1, 2] or col not in [0, 1, 2]:
                    print("Invalid input! Enter numbers between 0 and 2.")
                    continue

                if board[row][col] == " ":
                    board[row][col] = "X"
                else:
                    print("Spot is already taken! Choose another spot.")
                    continue
            except ValueError:
                print("Please enter valid integer numbers for row and column.")
                continue
        else:
            # Machine turn (Random choice)
            print("Computer's turn (O)...")
            empty_cells = [(r, c) for r in range(3) for c in range(3) if board[r][c] == " "]
            if empty_cells:
                row, col = random.choice(empty_cells)
                board[row][col] = "O"

        # Check if current player won
        if check_win(board, current_player):
            print_board(board)
            if current_player == "X":
                print("Congratulations! You win!")
            else:
                print("Computer wins! Better luck next time.")
            break

        # Check if it's a tie
        if check_tie(board):
            print_board(board)
            print("It's a tie!")
            break

        # Switch turn
        current_player = "O" if current_player == "X" else "X"

# Run the game
if __name__ == "__main__":
    play_game()
```
