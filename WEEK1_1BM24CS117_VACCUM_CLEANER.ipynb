```python
import random
import time

class VacuumEnvironment:
    def __init__(self):
        # 0 = Clean, 1 = Dirty
        self.location_status = {
            'A': random.choice([0, 1]),
            'B': random.choice([0, 1])
        }
        # Randomly place the agent in Room A or Room B
        self.agent_location = random.choice(['A', 'B'])

    def display_status(self):
        print(f"Environment Status -> Room A: {'DIRTY' if self.location_status['A'] else 'CLEAN'} | "
              f"Room B: {'DIRTY' if self.location_status['B'] else 'CLEAN'}")
        print(f"Agent is currently in: Room {self.agent_location}")


class ReflexVacuumAgent:
    def __init__(self):
        self.cost = 0  # Track path cost/actions performed

    def act(self, environment):
        print("--- Starting Simulation ---")
        environment.display_status()

        # The agent will continue until both rooms are clean
        while environment.location_status['A'] == 1 or environment.location_status['B'] == 1:
            current_loc = environment.agent_location
            current_status = environment.location_status[current_loc]

            time.sleep(1) # Visual pause between steps

            # Reflex Condition-Action Rules
            if current_status == 1:  # If location is Dirty
                print(f"[ACTION] Room {current_loc} is DIRTY. Sucking dirt...")
                environment.location_status[current_loc] = 0  # Clean the room
                self.cost += 1
                print(f"[STATUS] Room {current_loc} is now CLEAN.")

            else:  # If location is Clean, move to the other room
                if current_loc == 'A':
                    print("[ACTION] Room A is CLEAN. Moving RIGHT to Room B...")
                    environment.agent_location = 'B'
                else:
                    print("[ACTION] Room B is CLEAN. Moving LEFT to Room A...")
                    environment.agent_location = 'A'
                self.cost += 1

            environment.display_status()

        print("\n--- Goal Reached! All rooms are clean. ---")
        print(f"Total actions performed (Path Cost): {self.cost}")


# --- Execution Flow ---
if __name__ == "__main__":
    # Initialize environment and the reflex agent
    env = VacuumEnvironment()
    agent = ReflexVacuumAgent()

    # Run the agent in the environment
    agent.act(env)
```
